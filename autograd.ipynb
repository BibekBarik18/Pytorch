```python
import torch
```

```python
x=torch.randn(3,requires_grad=True)
```

```python
print(x)
y=x+x+2
print(y)
z=y*y*2
z=z.mean()
print(z)
# z.backward()
print(x.grad)
```

```python
# x.requires_grad_(False)
# y=x.detach()
with torch.no_grad():
    y=x+2
    
print(y)
```

```python
weights=torch.ones(4,requires_grad=True)

for epoch in range(2):
    model=(weights*3).sum()
    print(model)
    model.backward()
    print(weights.grad)
    weights.grad.zero_()
```
